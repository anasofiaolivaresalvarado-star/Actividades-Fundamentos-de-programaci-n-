## Introducción

En esta actividad voy a desarrollar un programa en Python para crear una tabla de Pitágoras del 1 al 10 utilizando una matriz, la cual estará organizada en filas y columnas para almacenar los resultados de las multiplicaciones.

Para construirla, utilizaré dos ciclos `for` anidados y el método `append()`. También crearé funciones que me permitan mostrar la tabla de manera ordenada y consultar los resultados directamente desde la matriz, sin tener que realizar nuevamente las multiplicaciones.

Además, el programa solicitará dos números al usuario y comprobará que se encuentren dentro del rango del 1 al 10. Con esta actividad pondré en práctica el uso de listas, matrices, ciclos, funciones y validación de datos.

```python
tabla = []

for renglon in range(1, 11):
    fila = []

    for columna in range(1, 11):
        fila.append(renglon * columna)

    tabla.append(fila)

print("La matriz de 10 x 10 se creó correctamente.")
```

## Explicación del código

Primero, creo una lista vacía llamada `tabla`, donde voy a guardar los resultados de las multiplicaciones.

Después, utilizo un ciclo `for` para recorrer los números del 1 al 10 y crear una lista llamada `fila` en cada repetición.

Dentro de este ciclo, utilizo otro `for` para recorrer las columnas del 1 al 10. Multiplico el número del renglón por el de la columna y guardo cada resultado utilizando `append()`.

Al terminar cada fila, utilizo `tabla.append(fila)` para agregarla a la matriz principal.

Finalmente, utilizo `print()` para mostrar un mensaje que confirma que la matriz de 10 × 10 se creó correctamente.

De esta manera, obtengo una matriz que contiene los resultados de las tablas de multiplicar del 1 al 10.

```python
def imprimir_tabla(tabla):
    for fila in tabla:
        for numero in fila:
            print(numero, end="\t")
        print()


def consultar_producto(tabla, renglon, columna):
    return tabla[renglon - 1][columna - 1]


print("Tabla de Pitágoras:")
imprimir_tabla(tabla)
```

### Explicación del código

Primero, defino la función `imprimir_tabla(tabla)`, que recibe la matriz que creé anteriormente y se encarga de mostrar todos sus valores de manera ordenada.

Dentro de esta función utilizo dos ciclos `for`. El primero recorre cada fila de la matriz y el segundo recorre los números que contiene cada fila. Después utilizo `print(numero, end="\t")` para mostrar los números separados por tabulaciones y `print()` para pasar a la siguiente fila. Esta función solamente muestra la información y no necesita utilizar `return`.

Después, defino la función `consultar_producto(tabla, renglon, columna)`, que recibe la matriz y dos números para localizar el resultado que necesito. Utilizo `return tabla[renglon - 1][columna - 1]` para obtener el valor de la posición indicada. Resto 1 a los números porque en Python las posiciones de las listas comienzan desde 0.

Finalmente, utilizo `print()` para mostrar el título de la tabla y llamo a la función `imprimir_tabla(tabla)` para visualizar todos los resultados.

```python
print("CONSULTA DE LA TABLA DE PITÁGORAS")

# Solicito el primer factor
renglon = int(input("Ingresa el primer factor (1-10): "))

# Verifico que el primer factor sea válido
while renglon < 1 or renglon > 10:
    print("El número está fuera del rango. Intenta nuevamente.")
    renglon = int(input("Ingresa el primer factor (1-10): "))

# Solicito el segundo factor
columna = int(input("Ingresa el segundo factor (1-10): "))

# Verifico que el segundo factor sea válido
while columna < 1 or columna > 10:
    print("El número está fuera del rango. Intenta nuevamente.")
    columna = int(input("Ingresa el segundo factor (1-10): "))

# Consulto el resultado en la matriz
resultado = consultar_producto(tabla, renglon, columna)

# Muestro el resultado
print("El resultado es:")
print(renglon, "x", columna, "=", resultado)
```

### Explicación del código

Este código permite consultar un resultado de la tabla de Pitágoras mediante dos factores ingresados por el usuario.

1. **Mostrar el título:** Se utiliza `print()` para mostrar el nombre del programa.

2. **Solicitar el primer factor:** Con `input()` se pide al usuario un número y con `int()` se convierte a entero.

3. **Validar el primer factor:** Se utiliza un ciclo `while` para comprobar que el número esté entre 1 y 10. Si está fuera del rango, se solicita nuevamente.

4. **Solicitar el segundo factor:** Se pide otro número, que también se convierte a entero.

5. **Validar el segundo factor:** Otro ciclo `while` verifica que el segundo número esté dentro del rango permitido.

6. **Consultar el resultado:** Se llama a la función `consultar_producto()` para obtener el producto de ambos factores utilizando la tabla almacenada en `tabla`.

7. **Mostrar el resultado:** Finalmente, se utiliza `print()` para mostrar la multiplicación y su resultado en pantalla.

**En conclusión:** El programa permite consultar multiplicaciones de la tabla de Pitágoras, validando que los dos números ingresados estén entre 1 y 10 antes de mostrar el resultado.
