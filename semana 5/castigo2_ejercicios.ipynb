## Ejercicio 1 — Bitácora de la semana

```python
mi_dias = ('Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes', 'Sábado', 'Domingo')
mi_km = (120, 85, 0, 210, 95, 0, 150)

def mi_resumen(mi_km):
    mi_total = 0
    mi_dias_rodados = 0
    mi_mayor = 0
    mi_indice_mayor = 0
    for mi_i in range(len(mi_km)):
        mi_total += mi_km[mi_i]
        if mi_km[mi_i] > 0:
            mi_dias_rodados += 1
        if mi_km[mi_i] > mi_mayor:
            mi_mayor = mi_km[mi_i]
            mi_indice_mayor = mi_i
    mi_promedio = round(mi_total / mi_dias_rodados, 1)
    return (mi_total, mi_dias_rodados, mi_promedio, mi_indice_mayor)
mi_total, mi_dias_rodados, mi_promedio, mi_indice = mi_resumen(mi_km)
print('Total:', mi_total, 'km')
print('Días con rodada:', mi_dias_rodados)
print('Promedio por día rodado:', mi_promedio, 'km')
print('Día más largo:', mi_dias[mi_indice], '(', mi_km[mi_indice], 'km)')
```

## Ejercicio 2 — La votación del festival

```python
mi_votos = [('Metallica', 48), ('Slayer', 22), ('Megadeth', 35), ('Anthrax', 15), ('Iron Maiden', 40)]

def mi_promedio_votos(mi_lista):
    mi_suma = 0
    for mi_banda, mi_cantidad in mi_lista:
        mi_suma += mi_cantidad
    return mi_suma / len(mi_lista)
mi_ganadora = mi_votos[0]
for mi_banda, mi_cantidad in mi_votos:
    if mi_cantidad > mi_ganadora[1]:
        mi_ganadora = (mi_banda, mi_cantidad)
mi_promedio = mi_promedio_votos(mi_votos)
print('Ganador:', mi_ganadora[0], 'con', mi_ganadora[1], 'votos')
print('Promedio:', mi_promedio)
print('Arriba del promedio:')
for mi_banda, mi_cantidad in mi_votos:
    if mi_cantidad > mi_promedio:
        print('-', mi_banda, '(', mi_cantidad, ')')
```

## Ejercicio 3 — Tramos de la ruta a la Sierra

```python
mi_paradas = (('Querétaro', 0), ('San Juan del Río', 52), ('Tequisquiapan', 74), ('Bernal', 125), ('Jalpan', 260))
mi_mayor = 0
mi_inicio_mayor = ''
mi_fin_mayor = ''
for mi_i in range(len(mi_paradas) - 1):
    mi_nombre1, mi_km1 = mi_paradas[mi_i]
    mi_nombre2, mi_km2 = mi_paradas[mi_i + 1]
    mi_distancia = mi_km2 - mi_km1
    print(mi_nombre1, '→', mi_nombre2 + ':', mi_distancia, 'km')
    if mi_distancia > mi_mayor:
        mi_mayor = mi_distancia
        mi_inicio_mayor = mi_nombre1
        mi_fin_mayor = mi_nombre2
print('Tramo más largo:', mi_inicio_mayor, '→', mi_fin_mayor, '(', mi_mayor, 'km)')
```

## Ejercicio 4 — ¿Con chamarra o sin chamarra?

```python
mi_temps = (8, 14, 22, 27, 31, 19, 11)

def mi_clasificar(mi_t):
    if mi_t < 12:
        return 'frío'
    elif mi_t <= 24:
        return 'templado'
    else:
        return 'calor'
mi_etiquetas = []
mi_contador = {'frío': 0, 'templado': 0, 'calor': 0}
for mi_temperatura in mi_temps:
    mi_etiqueta = mi_clasificar(mi_temperatura)
    mi_etiquetas.append(mi_etiqueta)
    mi_contador[mi_etiqueta] += 1
mi_predomina = ''
for mi_etiqueta, mi_cantidad in mi_contador.items():
    if mi_predomina == '' or mi_cantidad > mi_contador[mi_predomina]:
        mi_predomina = mi_etiqueta
print(mi_etiquetas)
print(mi_contador)
print('Predomina:', mi_predomina)
```

## Ejercicio 5 — ¿Cuánto dura el show?

```python
mi_setlist = (('Intro', 95), ('Riff asesino', 245), ('Balada', 302), ('Solo', 128), ('Encore', 409))

def mi_formato(mi_seg):
    mi_minutos = mi_seg // 60
    mi_segundos = mi_seg % 60
    if mi_segundos < 10:
        return f'{mi_minutos}:0{mi_segundos}'
    return f'{mi_minutos}:{mi_segundos}'
mi_total = 0
for mi_cancion, mi_segundos in mi_setlist:
    print(mi_cancion + ':', mi_formato(mi_segundos))
    mi_total += mi_segundos
print('Duración total:', mi_formato(mi_total))
```

## Ejercicio 7 — Boleta del grupo

```python
mi_notas = {'Ana': [90, 85, 78], 'Luis': [60, 72, 65], 'Mía': [100, 95, 98], 'Beto': [70, 68, 74]}

def mi_promedio(mi_lista):
    mi_suma = 0
    for mi_nota in mi_lista:
        mi_suma += mi_nota
    return round(mi_suma / len(mi_lista), 1)
mi_promedios = {}
for mi_alumno, mi_calificaciones in mi_notas.items():
    mi_promedios[mi_alumno] = mi_promedio(mi_calificaciones)
for mi_alumno, mi_calificacion in mi_promedios.items():
    if mi_calificacion >= 70:
        mi_estado = 'Aprobado'
    else:
        mi_estado = 'Reprobado'
    print(mi_alumno + ':', mi_calificacion, '-', mi_estado)
mi_mejor = ''
for mi_alumno, mi_calificacion in mi_promedios.items():
    if mi_mejor == '' or mi_calificacion > mi_promedios[mi_mejor]:
        mi_mejor = mi_alumno
print('Mejor promedio:', mi_mejor)
```

## Ejercicio 10 — Traductor de mensajes de WhatsApp

```python
mi_abrev = {'q': 'que', 'xq': 'porque', 'tmb': 'también', 'pq': 'porque', 'bn': 'bien', 'tqm': 'te quiero mucho'}

def mi_expandir(mi_mensaje):
    mi_palabras = mi_mensaje.lower().split()
    mi_nuevas = []
    for mi_palabra in mi_palabras:
        mi_nuevas.append(mi_abrev.get(mi_palabra, mi_palabra))
    return ' '.join(mi_nuevas)
print(mi_expandir('Q onda tmb voy al concierto xq tocan bn'))
print(mi_expandir('tqm Mamá'))
```

## Ejercicio 13 — Capturas sucias

`" 6 "` sí se convierte porque `int()` ignora los espacios al inicio y al final; `"8.5"` no se convierte porque tiene un punto decimal y `int()` requiere un número entero.

```python
mi_capturas = ['10', '7', 'x', '', '9', '8.5', ' 6 ']
mi_validas = []
mi_invalidas = 0
for mi_captura in mi_capturas:
    try:
        mi_numero = int(mi_captura)
        mi_validas.append(mi_numero)
    except ValueError:
        mi_invalidas += 1
mi_promedio = round(sum(mi_validas) / len(mi_validas), 2)
print('Válidas:', mi_validas)
print('Inválidas:', mi_invalidas)
print('Promedio:', mi_promedio)
```

## Ejercicio 15 — Cupones de descuento

```python
mi_cupones = {'METAL10': 10, 'MOTO20': 20, 'ROCK15': 15}

def mi_aplicar(mi_total, mi_codigo):
    mi_codigo = mi_codigo.strip().upper()
    try:
        mi_porcentaje = mi_cupones[mi_codigo]
        return mi_total - mi_total * mi_porcentaje / 100
    except KeyError:
        print('Cupón no válido')
        return mi_total
for mi_codigo in ['  moto20 ', 'GRATIS', 'metal10']:
    mi_resultado = mi_aplicar(1200, mi_codigo)
    print(repr(mi_codigo), '→ $' + str(mi_resultado))
```

## Ejercicio 18 — Contador de vocales

```python
mi_frase = 'el rock nunca va a morir en esta ciudad'

def mi_vocales(mi_palabra):
    mi_cantidad = 0
    for mi_letra in mi_palabra:
        if mi_letra in 'aeiou':
            mi_cantidad += 1
    return mi_cantidad
mi_palabras = mi_frase.split()
mi_conteo = {}
for mi_palabra in mi_palabras:
    mi_conteo[mi_palabra] = mi_vocales(mi_palabra)
print(mi_conteo)
mi_mayor = ''
for mi_palabra, mi_cantidad in mi_conteo.items():
    if mi_mayor == '' or mi_cantidad > mi_conteo[mi_mayor]:
        mi_mayor = mi_palabra
print('Más vocales:', mi_mayor, '(' + str(mi_conteo[mi_mayor]) + ')')
```

## Ejercicio 19 — Usuarios del laboratorio

```python
mi_alumnos = ['Bruce Dickinson', 'Brian Dickinson', 'Ana Lopez', 'Andrea Lopez', 'Angel Lopez']
mi_usados = []

def mi_usuario(mi_nombre):
    mi_partes = mi_nombre.lower().split()
    mi_base = mi_partes[0][0] + mi_partes[1]
    mi_candidato = mi_base
    mi_numero = 2
    while mi_candidato in mi_usados:
        mi_candidato = mi_base + str(mi_numero)
        mi_numero += 1
    mi_usados.append(mi_candidato)
    return mi_candidato
for mi_alumno in mi_alumnos:
    print(mi_alumno + ' →', mi_usuario(mi_alumno))
```
