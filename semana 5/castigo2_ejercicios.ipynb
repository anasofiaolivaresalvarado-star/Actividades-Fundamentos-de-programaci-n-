## Ejercicio 1 — Bitácora de la semana

```python
mi_dias = ('Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes', 'Sábado', 'Domingo')
mi_km = (120, 85, 0, 210, 95, 0, 150)

def mi_resumen(mi_km):
    mi_total = 0
    mi_dias_rodados = 0
    mi_mayor = 0
    mi_indice_mayor = 0
    for mi_i in range(len(mi_km)):
        mi_total += mi_km[mi_i]
        if mi_km[mi_i] > 0:
            mi_dias_rodados += 1
        if mi_km[mi_i] > mi_mayor:
            mi_mayor = mi_km[mi_i]
            mi_indice_mayor = mi_i
    mi_promedio = round(mi_total / mi_dias_rodados, 1)
    return (mi_total, mi_dias_rodados, mi_promedio, mi_indice_mayor)
mi_total, mi_dias_rodados, mi_promedio, mi_indice = mi_resumen(mi_km)
print('Total:', mi_total, 'km')
print('Días con rodada:', mi_dias_rodados)
print('Promedio por día rodado:', mi_promedio, 'km')
print('Día más largo:', mi_dias[mi_indice], '(', mi_km[mi_indice], 'km)')
```

## Ejercicio 2 — La votación del festival

```python
mi_votos = [('Metallica', 48), ('Slayer', 22), ('Megadeth', 35), ('Anthrax', 15), ('Iron Maiden', 40)]

def mi_promedio_votos(mi_lista):
    mi_suma = 0
    for mi_banda, mi_cantidad in mi_lista:
        mi_suma += mi_cantidad
    return mi_suma / len(mi_lista)
mi_ganadora = mi_votos[0]
for mi_banda, mi_cantidad in mi_votos:
    if mi_cantidad > mi_ganadora[1]:
        mi_ganadora = (mi_banda, mi_cantidad)
mi_promedio = mi_promedio_votos(mi_votos)
print('Ganador:', mi_ganadora[0], 'con', mi_ganadora[1], 'votos')
print('Promedio:', mi_promedio)
print('Arriba del promedio:')
for mi_banda, mi_cantidad in mi_votos:
    if mi_cantidad > mi_promedio:
        print('-', mi_banda, '(', mi_cantidad, ')')
```

## Ejercicio 3 — Tramos de la ruta a la Sierra

```python
mi_paradas = (('Querétaro', 0), ('San Juan del Río', 52), ('Tequisquiapan', 74), ('Bernal', 125), ('Jalpan', 260))
mi_mayor = 0
mi_inicio_mayor = ''
mi_fin_mayor = ''
for mi_i in range(len(mi_paradas) - 1):
    mi_nombre1, mi_km1 = mi_paradas[mi_i]
    mi_nombre2, mi_km2 = mi_paradas[mi_i + 1]
    mi_distancia = mi_km2 - mi_km1
    print(mi_nombre1, '→', mi_nombre2 + ':', mi_distancia, 'km')
    if mi_distancia > mi_mayor:
        mi_mayor = mi_distancia
        mi_inicio_mayor = mi_nombre1
        mi_fin_mayor = mi_nombre2
print('Tramo más largo:', mi_inicio_mayor, '→', mi_fin_mayor, '(', mi_mayor, 'km)')
```

## Ejercicio 4 — ¿Con chamarra o sin chamarra?

```python
mi_temps = (8, 14, 22, 27, 31, 19, 11)

def mi_clasificar(mi_t):
    if mi_t < 12:
        return 'frío'
    elif mi_t <= 24:
        return 'templado'
    else:
        return 'calor'
mi_etiquetas = []
mi_contador = {'frío': 0, 'templado': 0, 'calor': 0}
for mi_temperatura in mi_temps:
    mi_etiqueta = mi_clasificar(mi_temperatura)
    mi_etiquetas.append(mi_etiqueta)
    mi_contador[mi_etiqueta] += 1
mi_predomina = ''
for mi_etiqueta, mi_cantidad in mi_contador.items():
    if mi_predomina == '' or mi_cantidad > mi_contador[mi_predomina]:
        mi_predomina = mi_etiqueta
print(mi_etiquetas)
print(mi_contador)
print('Predomina:', mi_predomina)
```

## Ejercicio 5 — ¿Cuánto dura el show?

```python
mi_setlist = (('Intro', 95), ('Riff asesino', 245), ('Balada', 302), ('Solo', 128), ('Encore', 409))

def mi_formato(mi_seg):
    mi_minutos = mi_seg // 60
    mi_segundos = mi_seg % 60
    if mi_segundos < 10:
        return f'{mi_minutos}:0{mi_segundos}'
    return f'{mi_minutos}:{mi_segundos}'
mi_total = 0
for mi_cancion, mi_segundos in mi_setlist:
    print(mi_cancion + ':', mi_formato(mi_segundos))
    mi_total += mi_segundos
print('Duración total:', mi_formato(mi_total))
```
