## ¿Qué hace mi programa?

En esta actividad hice un programa con un menú que tiene cuatro secciones: tuplas, diccionarios, excepciones y strings. Cada sección tiene su propia función y puedo regresar al menú hasta que elija la opción de salir.

## Parte A — Tuplas

En esta parte trabajo con una tupla de números. Muestro el tercer elemento, agrego dos números creando una nueva tupla, ordeno sus elementos y calculo la suma usando una función.

```python
def sumar_tupla(datos):
    total = 0

    for numero in datos:
        total += numero

    return total


def seccion_tuplas():
    numeros = (10, 20, 30, 40, 50)

    print("\n========== SECCIÓN DE TUPLAS ==========")
    print("Mi tupla original es:", numeros)
    print("El tercer elemento de mi tupla es:", numeros[2])

    numero1 = int(input("Escribe el primer número: "))
    numero2 = int(input("Escribe el segundo número: "))

    nueva_tupla = numeros + (numero1, numero2)

    print("Mi nueva tupla es:", nueva_tupla)

    lista_numeros = list(nueva_tupla)
    lista_numeros.sort()

    print("Mi lista ordenada es:", lista_numeros)

    resultado = sumar_tupla(nueva_tupla)

    print("La suma de mi nueva tupla es:", resultado)
```

## Parte B — Diccionarios

Aquí creo una agenda con nombres y teléfonos. Después agrego un contacto nuevo, muestro los nombres guardados y busco el teléfono de una persona. Si no existe, muestro un mensaje en lugar de provocar un error.

```python
def buscar_telefono(contactos, nombre):
    return contactos.get(nombre)


def seccion_diccionarios():
    contactos = {
        "Carlos": "4421234567",
        "Laura": "4429876543",
        "Pedro": "4425551234"
    }

    print("\n========== SECCIÓN DE DICCIONARIOS ==========")

    nombre_nuevo = input("Escribe el nombre del nuevo contacto: ")
    telefono_nuevo = input("Escribe su teléfono: ")

    contactos[nombre_nuevo] = telefono_nuevo

    print("\nNombres de mis contactos:")

    for nombre in contactos:
        print(nombre)

    nombre_buscar = input("\n¿Qué contacto quieres buscar? ")

    telefono = buscar_telefono(contactos, nombre_buscar)

    if telefono is not None:
        print("El teléfono de", nombre_buscar, "es:", telefono)
    else:
        print(nombre_buscar, "no está en tus contactos")
```

## Parte C — Excepciones

En esta sección pido dos números enteros y calculo su suma y división. Uso try y except para controlar cuando escribo letras, dejo un espacio vacío o intento dividir entre cero.

```python
def seccion_excepciones():
    print("\n========== SECCIÓN DE EXCEPCIONES ==========")

    try:
        numero1 = int(input("Escribe el primer número entero: "))
        numero2 = int(input("Escribe el segundo número entero: "))

        suma = numero1 + numero2
        division = numero1 / numero2

        print("La suma es:", suma)
        print("La división es:", division)

    except ValueError:
        print("Error: debes escribir números enteros y no dejar espacios vacíos.")

    except ZeroDivisionError:
        print("Error: no se puede dividir entre cero.")
```

## Parte D — Strings

En esta parte uso un mensaje de texto para mostrar cuántos caracteres tiene, convertirlo a mayúsculas y cambiar una palabra por otra. También hago una función que cuenta cuántas palabras tiene el mensaje.

```python
def contar_palabras(texto):
    palabras = texto.split()
    return len(palabras)


def seccion_strings():
    print("\n========== SECCIÓN DE STRINGS ==========")

    mensaje = "Me gusta aprender programación"

    print("Mi mensaje original es:", mensaje)
    print("La cantidad de caracteres es:", len(mensaje))
    print("Mi mensaje en mayúsculas es:", mensaje.upper())

    mensaje_nuevo = mensaje.replace("programación", "Python")

    print("Mi mensaje modificado es:", mensaje_nuevo)

    cantidad = contar_palabras(mensaje)

    print("La cantidad de palabras es:", cantidad)
```

## Parte E — Menú modular

En esta última parte uno todas las secciones en un menú. Uso un ciclo while para que las opciones aparezcan nuevamente después de cada actividad. También agrego una opción para salir y un mensaje cuando escribo una opción que no existe.

```python
from unittest.mock import patch


def menu():
    while True:
        print("\n======================================")
        print("             MENÚ MODULAR")
        print("======================================")
        print("1) Tuplas")
        print("2) Diccionarios")
        print("3) Excepciones")
        print("4) Strings")
        print("5) Salir")
        print("======================================")

        opcion = input("Elige una opción: ")

        if opcion == "1":
            seccion_tuplas()

        elif opcion == "2":
            seccion_diccionarios()

        elif opcion == "3":
            seccion_excepciones()

        elif opcion == "4":
            seccion_strings()

        elif opcion == "5":
            print("\nPrograma terminado. ¡Hasta luego!")
            break

        else:
            print("\nOpción inválida. Intenta nuevamente.")


# Estas respuestas me sirven para probar todas las opciones
respuestas = iter([
    "1", "60", "25",
    "2", "Sofia", "4421112233", "Carlos",
    "2", "Ana", "4423334455", "Juan",
    "3", "12", "3",
    "3", "hola", "4",
    "3", "", "4",
    "3", "10", "0",
    "4",
    "9",
    "5"
])


# Muestro las respuestas para que se vea qué estoy probando
def entrada_prueba(mensaje=""):
    respuesta = next(respuestas)
    print(mensaje + respuesta)
    return respuesta


# Ejecuto el menú una sola vez con todas las pruebas
with patch("builtins.input", side_effect=entrada_prueba):
    menu()
```
