```python
import pdb

total = 0.0
registro   = 0

while True:
    print("\n=== KLUU FOOD SERVICES ===")
    print("1. Registrar despacho")
    print("2. Ver resumen")
    print("3. Salir")
    
    opcion = input("Elige una opción (1-3): ")
    
    if opcion == "1":
        precio = float(input("Ingresa el precio de el producto: "))
        cantidad = int(input("Ingresa la cantidad que desee comprar: "))
        
        subtotal = precio * cantidad
        
        if cantidad >= 100:
            descuento = subtotal * 0.15
        elif cantidad >= 50:
            descuento = subtotal * 0.10
        elif cantidad >= 25:
            descuento = subtotal * 0.05
        else:
            descuento = 0.0
            
        total_venta = subtotal - descuento
        
        
        total += total_venta
        registro += 1
        
        print("ELtotal a pagar: $", round(total_venta, 2))
        
    elif opcion == "2":
        print("--- Resumen ---")
        print("Total de despachos:", registro)
        print("Monto acumulado: $", round(total, 2))
        
    elif opcion == "3":
        print("Gracias por usar KLUU FOOD SERVICES. ¡Hasta luego!")
        break
    else:
        print("Por favor elige una opción válida (1-3).")
        
    pdb.set_trace()
```
