## Ejercicio 2. Modelo Entrada–Proceso–Salida (EPS)

En este ejercicio voy a identificar los datos de entrada, las operaciones que se realizan y el resultado que debe mostrar un programa.

**Problema:** Un cliente realiza una compra en una papelería por un importe de $500 y recibe un descuento del 10 %. El objetivo es calcular el descuento y el total que debe pagar.

**Entrada:**

- Subtotal de la compra: $500.
- Porcentaje de descuento: 10 %.

**Proceso:**

1. Multiplico el subtotal por el porcentaje de descuento.
2. Resto el descuento al subtotal de la compra.

**Salida esperada:**

- Descuento: $50.
- Total a pagar: $450.

```python
# Ejercicio 2: modelo Entrada-Proceso-Salida

subtotal_papeleria = 500
descuento_porcentaje = 0.10

importe_descuento = subtotal_papeleria * descuento_porcentaje
total_compra = subtotal_papeleria - importe_descuento

print("Subtotal: $", subtotal_papeleria)
print("Descuento: $", importe_descuento)
print("Total a pagar: $", total_compra)
```

## Ejercicio 4. Estructuras condicionales

En este ejercicio voy a practicar la toma de decisiones en Python mediante las instrucciones `if`, `elif` y `else`.

**Problema:** En una papelería se ofrece un descuento del 10 % a los estudiantes cuando el importe de su compra es de $150 o más.

**Entrada:**

- Importe de la compra.
- Indicación de si el cliente es estudiante.

**Proceso:**

1. Solicito el importe de la compra.
2. Verifico si el cliente es estudiante.
3. Compruebo si la compra alcanza los $150.
4. Si cumple ambas condiciones, calculo el descuento.
5. Si no cumple los requisitos, mantengo el importe original.

**Salida esperada:** El programa muestra el descuento aplicado y el total que debe pagar el cliente.

```python
# Ejercicio 4: estructuras condicionales

importe_compra = float(input("Ingresa el importe de la compra: "))
es_estudiante = input("¿El cliente es estudiante? (S/N): ").strip().upper()

if es_estudiante not in ("S", "N"):
    print("Respuesta no válida. Escribe S o N.")
elif es_estudiante == "S" and importe_compra >= 150:
    descuento = importe_compra * 0.10
    total_pagar = importe_compra - descuento
    print("Descuento aplicado: $", round(descuento, 2))
    print("Total después del descuento: $", round(total_pagar, 2))
else:
    descuento = 0
    total_pagar = importe_compra
    print("No aplica descuento.")
    print("Total a pagar: $", round(total_pagar, 2))
```

```python
# Pruebas automáticas de las condiciones

casos_prueba = [
    (195, "S"),
    (115, "S"),
    (195, "N")
]

for importe_compra, es_estudiante in casos_prueba:
    if es_estudiante == "S" and importe_compra >= 150:
        descuento = importe_compra * 0.10
    else:
        descuento = 0

    total_pagar = importe_compra - descuento

    print("--------------------")
    print("Importe de compra: $", importe_compra)
    print("Estudiante:", es_estudiante)
    print("Descuento: $", round(descuento, 2))
    print("Total: $", round(total_pagar, 2))
```

## Ejercicio 5. Ciclo while y validación de datos

En este ejercicio voy a utilizar el ciclo `while` para validar la cantidad de productos que desea comprar un cliente.

**Problema:** En una papelería no se pueden registrar cantidades negativas de productos. Si el usuario ingresa un número menor que cero, el programa debe solicitar nuevamente la cantidad.

**Entrada:** La cantidad de cuadernos que desea comprar.

**Proceso:**

1. Solicito la cantidad de cuadernos.
2. Compruebo si el número es negativo.
3. Mientras la cantidad sea negativa, vuelvo a solicitar el dato.
4. Cuando el valor sea válido, muestro la cantidad registrada y su costo.

**Salida esperada:** El programa acepta cantidades iguales o mayores que cero y rechaza los valores negativos.

```python
# Ejercicio 5: ciclo while

cantidad_productos = int(input("¿Cuántos cuadernos deseas comprar? "))

while cantidad_productos < 0:
    print("Error: la cantidad no puede ser negativa.")
    cantidad_productos = int(input("Ingresa una cantidad válida: "))

print("Cantidad registrada:", cantidad_productos)
print("Costo de los cuadernos: $", cantidad_productos * 35)
```

## Ejercicio 6. Acumuladores

En este ejercicio voy a practicar cómo sumar diferentes cantidades para obtener el total de ventas de una papelería.

**Problema:** Necesito acumular los importes de varios productos vendidos para conocer cuánto se obtuvo en total.

**Entrada:** Los importes de dos ventas.

**Proceso:** Utilizo una variable acumuladora para sumar cada uno de los importes registrados.

**Salida esperada:** El total acumulado de las ventas realizadas.

```python
# Ejercicio 6: acumulador de ventas

ventas_totales = 0

venta_cuadernos = 133.40
venta_materiales = 203.58

ventas_totales = ventas_totales + venta_cuadernos
ventas_totales = ventas_totales + venta_materiales

print("Venta de cuadernos: $", venta_cuadernos)
print("Venta de materiales: $", venta_materiales)
print("Ventas acumuladas: $", round(ventas_totales, 2))
```

## Ejercicio 7. Depuración de programas con PDB

En este ejercicio voy a practicar la depuración de un programa en Python mediante la herramienta `pdb`.

**Problema:** En una papelería necesito comprobar que el cálculo del subtotal de una compra sea correcto. Para ello, voy a detener temporalmente el programa y revisar los valores almacenados en sus variables.

**Entrada:**

- Precio de cada cuaderno: $35.
- Cantidad de cuadernos: 3.

**Proceso:**

1. Asigno el precio y la cantidad a sus variables.
2. Utilizo `pdb` para detener la ejecución.
3. Reviso los valores de las variables.
4. Continúo el programa y compruebo el cálculo.

**Salida esperada:** El programa muestra un precio de $35, una cantidad de 3 cuadernos y un subtotal de $105.

```python
# Ejercicio 7: Depuración de programas con PDB

import pdb

precio_cuaderno = 35
cantidad_cuadernos = 3

pdb.set_trace()

subtotal_compra = precio_cuaderno * cantidad_cuadernos

print("Precio por cuaderno: $", precio_cuaderno)
print("Cantidad de cuadernos:", cantidad_cuadernos)
print("Subtotal de la compra: $", subtotal_compra)
```

## Ejercicio 8. Ciclos for

En este ejercicio voy a practicar el ciclo `for` para repetir instrucciones una cantidad determinada de veces.

**Problema:** En una papelería quiero identificar cuántos números pares existen del 1 al 10 para practicar el recorrido de secuencias y el uso de contadores.

**Entrada:**

- Números enteros del 1 al 10.

**Proceso:**

1. Utilizo un ciclo `for` para recorrer los números del 1 al 10.
2. Compruebo cuáles son divisibles entre 2.
3. Cada vez que encuentro un número par, aumento un contador.
4. Muestro los números encontrados y la cantidad total.

**Salida esperada:** Los números pares son 2, 4, 6, 8 y 10. En total hay 5 números pares.

```python
# Ejercicio 8: ciclo for y contador

contador_pares = 0

print("Números pares del 1 al 10:")

for numero_actual in range(1, 11):
    if numero_actual % 2 == 0:
        print(numero_actual)
        contador_pares += 1

print("Cantidad total de números pares:", contador_pares)
```
