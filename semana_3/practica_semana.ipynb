Ejercicio 11: Suma con centinela 0

```python
suma = 0
numero = int(input("agrega un numero y pon 0 para terminar: "))

while numero != 0:
    if numero > 0:
        suma += numero
    numero = int(input("agrega otro numero (0 para terminar): "))

print("la suma es:", suma)
```

Ejercicio 12: Contar números pare

```python
pares = 0
numero = int(input("agrega un numero y pon negativo para terminar: "))

while numero >= 0:
    if numero % 2 == 0:
        pares += 1
    numero = int(input("agrega otro numero (negativo para terminar): "))

print("la cantidad de numeros pares es:", pares)
```

Ejercicio 13: Promedio con centinela -1

```python
suma = 0
cantidad = 0
calificacion = float(input("agrega una calificación y pon -1 para terminar: "))

while calificacion != -1:
    suma += calificacion
    cantidad += 1
    calificacion = float(input("agrega otra calificación y pon -1 para terminar: "))

if cantidad > 0:
    promedio = suma / cantidad
    print("el promedio es:", promedio)
else:
    print("no ingresaste ninguna calificacion")
```

Ejercicio 14: Menú simple con ciclo while

```python
opcion = 0

while opcion != 3:
    print("1. Sumar")
    print("2. Restar")
    print("3. Salir")
    opcion = int(input("elige una opcion: "))

    if opcion == 1:
        n1 = float(input("primero numero: "))
        n2 = float(input("segundo numero: "))
        print(f"Resultado: {n1 + n2}")
    elif opcion == 2:
        n1 = float(input("`primer numero: "))
        n2 = float(input("segundo numero: "))
        print(f"Resultado: {n1 - n2}")
    elif opcion == 3:
        print("se cierra el programa")
    else:
        print("opcion invalida, intenta de nuevo")
```

Ejercicio 15: Tabla de multiplicar con while

```python
numero = int(input("agregaa un numero: "))
i = 1

while i <= 10:
    print(f"{numero} x {i} = {numero * i}")
    i += 1
```
