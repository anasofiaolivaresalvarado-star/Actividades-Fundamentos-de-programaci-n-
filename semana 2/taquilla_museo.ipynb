Mi programa sirve para calcular el costo de los boletos de un museo según la edad y los descuentos que cada visitante tiene. Al final, muestra un resumen con el total de dinero recaudado y boletos vendidos. Mi programa tiene diferentes condiciones dependiendo el tipo de visitante que sea el que desea comprar el boleto. 

PARTE B

```python
precio_menor = 30.0
precio_adulto = 45.0

descuento_estudiante = 0.10
descuento_profesor = 0.10
descuento_adulto_mayor = 0.12

total_boletos = 0
total_recaudado = 0.0
total_descuentos = 0.0

print("=== TAQUILLA DEL MUSEO ===")

while True: 
    cantidad_entrada = input("¿cuántos visitantes desean comprar boletos (o 'fin' para cerrar la taquilla): ")
    
    if cantidad_entrada.lower() == 'fin':
        print("la taquilla se ha cerrado, gracias por entrar al programa")
        break

    cantidad_de_visitantes = int(cantidad_entrada)

    for i in range(1, cantidad_de_visitantes + 1):
        print(f"visitante {i}")

        visitante_edad_input = input(f"ingrese la edad del visitante {i} (o 'fin' para cerrar): ")
        
        if visitante_edad_input.lower() == 'fin':
            print("la taquilla se ha cerrado, gracias por entrar al programa")
            break
            
        visitante_edad = int(visitante_edad_input) 

        if visitante_edad < 0 or visitante_edad > 110:
            print("por favor ingrese una edad válida (0-110)")
            continue

        
        if visitante_edad < 3:   
            costo_base = 0.0
            tipo = "menor de 3"
            descuento = 0.0
            precio_por_persona = 0.0
            print("costo $0, entra gratis")
            
        elif visitante_edad <= 17:
            costo_base = precio_menor
            tipo = "menor de edad"
            descuento = 0.0 
            precio_por_persona = costo_base
            
        else:
            costo_base = precio_adulto
            
            
            opcion_tipo = input("ingrese el tipo de visitante: 1) estudiante  2) profesor  3) ninguno: ")
            
            if visitante_edad >= 60:
                tipo = "adulto mayor"
                descuento = costo_base * descuento_adulto_mayor
            else:
                if opcion_tipo == "1":
                    tipo = "estudiante"
                    descuento = costo_base * descuento_estudiante
                elif opcion_tipo == "2":
                    tipo = "profesor"
                    descuento = costo_base * descuento_profesor
                else:
                    tipo = "ninguno"
                    descuento = 0.0
                
            precio_por_persona = costo_base - descuento
        
        total_boletos += 1
        total_descuentos += descuento
        total_recaudado += precio_por_persona

        print(f"visitante {total_boletos} | edad: {visitante_edad} | tipo: {tipo} | base: ${costo_base:.2f} | descuento: ${descuento:.2f} | total a pagar: ${precio_por_persona:.2f}")
    
    if visitante_edad_input.lower() == 'fin':
        break
        
    visitantes_extra = input("¿quiere agregar mas visitantes? (escriba 'fin' para cerrar): ")
    if visitantes_extra.lower() == 'fin':
        print("perfecto!, gracias por entrar al programa, hasta luego")
        break

print("\n=== CORTE DE CAJA ===")
print(f"visitantes cobrados: {total_boletos}")
print(f"total en descuentos: ${total_descuentos:.2f}")
print(f"TOTAL A PAGAR: ${total_recaudado:.2f}")
```
